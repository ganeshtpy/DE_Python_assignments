```python

#Write a program that converts 5000 seconds into hours, minutes, and remaining seconds.

x=int(input("Enter the conversion value in seconds: "))
Hours=int(x/3600)
Y=x%3600
A= Y/60
Minutes= int(A)
seconds=round((A-Minutes)*60)

# Seconds=(5000-)
print(Hours, Minutes, seconds)
```
