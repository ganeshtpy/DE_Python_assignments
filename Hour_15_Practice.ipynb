```python
#Write a program to count the number of vowels in a user-provided string using a for loop.

x=input("Enter the word:")
vowel=0
for chr in x:
  if chr in 'aeiou':
    vowel += 1
print(vowel)
```

```python
#Print numbers from 10 down to 1 using range()
for i in range(10,0,-1):
  print(i,end=",")
```

```python
#Write a program that calculates the factorial of a number entered by the user.
num = int(input("Enter a number: "))

factrl = 1

for i in range(num, 0, -1):
    factrl *= i

print(factrl)
```

```python
#Print a pattern of stars: 1 star in first line, 2 in second, up to 5 stars in fifth line (using loop).

for i in range(1, 6):
    print("*" * i)
```

```python
#Given a list of names (defined as string space-separated or loop on letters), count the occurrences of letter 'e'.

names = ("Edukron gives Data Engineering Course").lower()
count = 0
for char in names:
    if char.lower() == 'e':
         count += 1
print(count)
```
